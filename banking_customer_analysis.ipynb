# Banking Customer Analysis

This notebook explores a synthetic retail-banking customer dataset.

The objective is to practice Python data analysis while answering practical banking questions about customer behaviour, balances, digital adoption and churn.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("banking_customers.csv")
df.head()
```

## 1. Understand the dataset

```python
print("Rows and columns:", df.shape)
print("\nMissing values:")
print(df.isna().sum())
print("\nData types:")
print(df.dtypes)
```

## 2. Customer overview

```python
summary = {
    "Customers": len(df),
    "Average Age": round(df["Age"].mean(), 1),
    "Average Balance (PKR)": round(df["AccountBalancePKR"].mean()),
    "Average Credit Score": round(df["CreditScore"].mean()),
    "Average Monthly Transactions": round(df["MonthlyTransactions"].mean(), 1)
}
pd.Series(summary)
```

This gives a quick picture of the typical customer in the sample before looking at more specific behaviour.

## 3. Overall churn rate

```python
churn_rate = (df["Churned"].eq("Yes").mean() * 100)
print(f"Overall churn rate: {churn_rate:.1f}%")
```

## 4. Churn by customer activity

```python
churn_by_activity = (
    df.groupby("ActiveMember")["Churned"]
      .apply(lambda x: (x == "Yes").mean() * 100)
)

print(churn_by_activity)

ax = churn_by_activity.plot(kind="bar", figsize=(7, 4))
ax.set_title("Churn Rate by Customer Activity")
ax.set_xlabel("Active Member")
ax.set_ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Interpretation:** In this synthetic dataset, inactive customers churn more often than active customers.  
Inactive churn: **31.6%** vs active churn: **12.5%**.

For a bank, this can justify early engagement or retention campaigns for customers whose activity drops.

## 5. Digital banking and churn

```python
digital_churn = (
    df.groupby("DigitalBankingUser")["Churned"]
      .apply(lambda x: (x == "Yes").mean() * 100)
)

digital_churn
```

**Interpretation:** Customers using digital banking had a churn rate of **16.2%**, while non-digital users had a churn rate of **23.6%** in this sample.

This does not prove that digital banking causes lower churn, but it suggests digital engagement is worth investigating.

## 6. Churn by number of products

```python
churn_by_products = (
    df.groupby("NumProducts")["Churned"]
      .apply(lambda x: (x == "Yes").mean() * 100)
)

print(churn_by_products)

ax = churn_by_products.plot(kind="bar", figsize=(7, 4))
ax.set_title("Churn Rate by Number of Products")
ax.set_xlabel("Number of Products")
ax.set_ylabel("Churn Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

**Interpretation:** The highest churn rate appears in the **1-product** group at **27.1%**.

A bank could study whether customers with fewer relationships need better onboarding, cross-sell opportunities or more relevant product recommendations.

## 7. Average account balance by city

```python
balance_by_city = (
    df.groupby("City")["AccountBalancePKR"]
      .mean()
      .sort_values(ascending=False)
)

print(balance_by_city)

ax = balance_by_city.plot(kind="bar", figsize=(8, 4))
ax.set_title("Average Account Balance by City")
ax.set_xlabel("City")
ax.set_ylabel("Average Balance (PKR)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

**Interpretation:** **Karachi** has the highest average account balance in this synthetic sample.

In a real bank, geographic differences could be explored further alongside income, product mix, branch footprint and customer demographics.
